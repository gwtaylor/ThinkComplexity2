## Complete graph

To make a complete graph, we use a generator function that iterates through all pairs of nodes.

```python
# GWT: note the generator here
def all_pairs(nodes):
    for i, u in enumerate(nodes):
        for j, v in enumerate(nodes):
            if i < j:
                yield u, v
```

`make_complete_graph` makes a `Graph` with the given number of nodes and edges between all pairs of nodes.

```python
def make_complete_graph(n):
    G = nx.Graph()
    nodes = range(n)
    G.add_nodes_from(nodes)
    G.add_edges_from(all_pairs(nodes))
    return G
```

Here's a complete graph with 10 nodes:

```python
complete = make_complete_graph(10)
len(complete.nodes())
```

And here's what it looks like.

```python
nx.draw_circular(complete, 
                 node_color=COLORS[2], 
                 node_size=1000, 
                 with_labels=True)
plt.savefig('chap02-3.pdf')
```

The `neighbors` method returns a list of neighbors for a given node.

```python
complete.neighbors(0)
```

## Random graphs

Next we'll make a random graph where the probability of an edge between each pair of nodes is $p$.

The helper function `flip` returns True with probability `p` and False with probability `1-p`

```python
from numpy.random import random

def flip(p):
    return random() < p
```

`random_pairs` is a generator function that enumerates all possible pairs of nodes and yields each one with probability `p` 

```python
def random_pairs(nodes, p):
    for i, u in enumerate(nodes):
        for j, v in enumerate(nodes):
            if i<j and flip(p):
                yield u, v
```

`make_random_graph` makes an ER graph where the probability of an edge between each pair of nodes is `p`.

```python
def make_random_graph(n, p):
    G = nx.Graph()
    nodes = range(n)
    G.add_nodes_from(nodes)
    G.add_edges_from(random_pairs(nodes, p))
    return G
```

Here's an example with `n=10` and `p=0.3`

```python
random_graph = make_random_graph(10, 0.3)
len(random_graph.edges())
```

And here's what it looks like:

```python
nx.draw_circular(random_graph, 
                 node_color=COLORS[3], 
                 node_size=1000, 
                 with_labels=True)
plt.savefig('chap02-4.pdf')
```

## Connectivity

To check whether a graph is connected, we'll start by finding all nodes that can be reached, starting with a given node:

```python
# GWT: note that stack is implemented as a Python list
def reachable_nodes(G, start):
    seen = set()
    stack = [start]
    while stack:
        node = stack.pop()
        if node not in seen:
            seen.add(node)
            stack.extend(G.neighbors(node))
    return seen
```

In the complete graph, starting from node 0, we can reach all nodes:

```python
reachable_nodes(complete, 0)
```

In the random graph we generated, we can also reach all nodes (but that's not always true):

```python
reachable_nodes(random_graph, 0)
```

We can use `reachable_nodes` to check whether a graph is connected:

```python
# GWT: note here the creation of an iterator over nodes
# And next() returns the next item from the iterator
# Here, it's an arbitrary node
def is_connected(G):
    start = next(G.nodes_iter())
    reachable = reachable_nodes(G, start)
    return len(reachable) == len(G)
```

Again, the complete graph is connected:

```python
is_connected(complete)
```

But if we generate a random graph with a low value of `p`, it's not:

```python
random_graph = make_random_graph(10, 0.1)
len(random_graph.edges())
```

```python
is_connected(random_graph)
```

## Probability of connectivity

Now let's estimate the probability that a randomly-generated ER graph is connected.

This function takes `n` and `p`, generates `iters` graphs, and returns the fraction of them that are connected.

```python
# GWT: Note, this is called a Monte Carlo estimate
def prob_connected(n, p, iters=100):
    count = 0
    for i in range(iters):
        random_graph = make_random_graph(n, p)
        if is_connected(random_graph):
            count += 1
    return count/iters
```

With `n=10` and `p=0.3`, the probability of being connected is about 65%.

```python
n = 10
prob_connected(n, 0.3, iters=10000)
```

According to Erdos and Renyi, the critical value of `p` for `n=10` is about 0.23. 

```python
pstar = np.log(n) / n
pstar
```

So let's plot the probability of connectivity for a range of values for `p`

```python
# GWT: logspace works in base 10 by default
# So this plots log uniform-spaced values between 10**(-1.3) and 1
ps = np.logspace(-1.3, 0, 11)
ps
```

I'll estimate the probabilities with `iters=1000`

```python
ys = [prob_connected(n, p, 1000) for p in ps]

for p, y in zip(ps, ys):
    print(p, y)
```

And then plot them, adding a vertical line at the computed critical value

```python
import thinkplot

thinkplot.vlines([pstar], 0, 1, color='gray')
thinkplot.plot(ps, ys)
thinkplot.config(xlabel='p', ylabel='prob connected', xscale='log', xlim=[ps[0], ps[-1]])
plt.savefig('chap02-5.pdf')
```

We can run the same analysis for a few more values of `n`.

```python
ns = [30, 100, 300]
ps = np.logspace(-2.5, 0, 11)

thinkplot.preplot(len(ns))

# GWT: This takes a bit of time!
for n in ns:
    pstar = np.log(n) / n
    thinkplot.vlines([pstar], 0, 1, color='gray')

    ys = [prob_connected(n, p) for p in ps]
    thinkplot.plot(ps, ys, label='n=%d' % n)

thinkplot.config(xlabel='p', ylabel='prob connected',
                 xscale='log', xlim=[ps[0], ps[-1]],
                 loc='upper left')
plt.savefig('chap02-6.pdf')
```

As `n` increases, the critical value gets smaller and the transition gets more abrupt.
